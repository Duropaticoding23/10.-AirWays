```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import time

print("Libraries imported successfully!")
```

```python
train_path = "/content/train.csv"

df = pd.read_csv(train_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
display(df.head())
```

```python
print("Number of columns:", len(df.columns))

print("\nFirst 10 columns:")
print(df.columns[:10].tolist())

print("\nLast 10 columns:")
print(df.columns[-10:].tolist())
```

```python
print("Target column exists:", "target" in df.columns)
```

```python
print(df["target"].value_counts())
```

```python
print(df["target"].value_counts(normalize=True) * 100)
```

```python
missing_values = df.isnull().sum().sum()

print("Total missing values:", missing_values)
```

```python
X = df.drop(columns=["ID_code", "target"])
y = df["target"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

print("train_test_split imported successfully!")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.naive_bayes import GaussianNB

nb_model = GaussianNB()

print("Gaussian Naive Bayes model created!")
```

```python
start_time = time.time()

nb_model.fit(X_train, y_train)

training_time = time.time() - start_time

print(f"Training time: {training_time:.4f} seconds")
```

```python
# Predicted class
y_pred = nb_model.predict(X_test)

# Probability of class 1
y_prob = nb_model.predict_proba(X_test)[:, 1]

print("Predictions completed!")
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    y_prob
)

print("Naive Bayes Results")
print("---------------------------")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-score  : {f1:.4f}")
print(f"ROC-AUC   : {roc_auc:.4f}")
print(f"Training Time: {training_time:.4f} seconds")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Class 0", "Class 1"]
)

disp.plot()

plt.title("Naive Bayes - Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```

```python
results = pd.DataFrame({
    "Algorithm": ["Gaussian Naive Bayes"],
    "Accuracy": [accuracy],
    "Precision": [precision],
    "Recall": [recall],
    "F1-score": [f1],
    "ROC-AUC": [roc_auc],
    "Training Time (seconds)": [training_time]
})

display(results)
```

```python
from sklearn.metrics import roc_curve

# Calculate ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# Plot
plt.figure(figsize=(7, 5))

plt.plot(
    fpr,
    tpr,
    label=f"Gaussian NB (AUC = {roc_auc:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Naive Bayes - ROC Curve")
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.metrics import classification_report

print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Class 0", "Class 1"],
        zero_division=0
    )
)
```

```python
results.to_csv(
    "/content/naive_bayes_results.csv",
    index=False
)

print("Results saved successfully!")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
import time

split_results = []

# Three train/test rules
splits = {
    "80/20": 0.20,
    "70/30": 0.30,
    "90/10": 0.10
}

for split_name, test_size in splits.items():

    # Split dataset
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_size,
        random_state=42,
        stratify=y
    )

    # Create model
    model = GaussianNB()

    # Training time
    start_time = time.time()
    model.fit(X_train, y_train)
    training_time = time.time() - start_time

    # Predictions
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_test, y_prob)

    split_results.append({
        "Split": split_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1,
        "ROC-AUC": roc_auc,
        "Training Time (sec)": training_time
    })

split_results_df = pd.DataFrame(split_results)

display(split_results_df)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    split_results_df["Split"],
    split_results_df["Accuracy"]
)

plt.xlabel("Train/Test Split")
plt.ylabel("Accuracy")
plt.title("Naive Bayes Accuracy for Different Train/Test Splits")

plt.ylim(0, 1)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-score",
    "ROC-AUC"
]

split_results_df.set_index("Split")[metrics].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Naive Bayes Performance for Different Train/Test Splits")
plt.xlabel("Train/Test Split")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.legend()
plt.show()
```
